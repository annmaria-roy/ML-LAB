```python
import numpy as np
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB,BernoulliNB
from sklearn.metrics import classification_report

categories=['comp.graphics','rec.sport.hockey','sci.med','talk.politics.mideast']
print("Fetching dataset.....")
train_data=fetch_20newsgroups(subset='train',categories=categories,remove=('headers','footers','quotes'))
test_data=fetch_20newsgroups(subset='test',categories=categories,remove=('headers','footers','quotes'))
print("Done")
```

```python
y_train = train_data.target
y_test = test_data.target

print(f"Loaded {len(train_data.data)} training documents "
      f"and {len(test_data.data)} test documents.")

# Multinomial NB - word count representation
mnb_vectorizer = CountVectorizer(
    stop_words='english',
    max_features=10000
)
```

```python
X_train_counts = mnb_vectorizer.fit_transform(train_data.data)
X_test_counts = mnb_vectorizer.transform(test_data.data)

# Bernoulli NB - binary word presence representation
bnb_vectorizer = CountVectorizer(
    stop_words='english',
    max_features=10000,
    binary=True
)
```

```python
X_train_binary = bnb_vectorizer.fit_transform(train_data.data)
X_test_binary = bnb_vectorizer.transform(test_data.data)

print("Vectorization strategies initialized and applied.")

# Train Multinomial Naïve Bayes
mnb = MultinomialNB(alpha=1.0)
mnb.fit(X_train_counts, y_train)

y_pred_mnb = mnb.predict(X_test_counts)

# Train Bernoulli Naïve Bayes
bnb = BernoulliNB(alpha=1.0)
bnb.fit(X_train_binary, y_train)

y_pred_bnb = bnb.predict(X_test_binary)
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score, classification_report

# Calculate evaluation metrics
results = {
    "Accuracy": [
        accuracy_score(y_test, y_pred_mnb),
        accuracy_score(y_test, y_pred_bnb)
    ],
    "Macro F1": [
        f1_score(y_test, y_pred_mnb, average='macro'),
        f1_score(y_test, y_pred_bnb, average='macro')
    ]
}

# Create a DataFrame to display the results nicely
df_results = pd.DataFrame(results, index=["Multinomial NB", "Bernoulli NB"])
print("Overall Results:")
print(df_results.round(4))
print("\n" + "="*60 + "\n")

# Display individual classification reports
print("Multinomial NB Classification Report:")
print(classification_report(y_test, y_pred_mnb, target_names=categories))
print("\n" + "="*60 + "\n")

print("Bernoulli NB Classification Report:")
print(classification_report(y_test, y_pred_bnb, target_names=categories))
```
